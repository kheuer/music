```python
(X_train, y_train, ids_train),(X_test, y_test, ids_test),(X_val, y_val, ids_val) = get_train_test_val_features_and_targets(df=df, feature_extractor=compute_spectral_features, splits=10, train_size=0.6, test_size=0.2)
```

```python
X_train_reshaped = X_train.reshape(X_train.shape[0], X_train.shape[1]*X_train.shape[2])
X_test_reshaped = X_test.reshape(X_test.shape[0], X_test.shape[1]*X_test.shape[2])
X_val_reshaped = X_val.reshape(X_val.shape[0], X_val.shape[1]*X_val.shape[2])
print(X_train_reshaped.shape)
```

```python
ref = create_random_forest(max_depth=None, min_samples_leaf=5, n_estimators=200)
ref.fit(X_train_reshaped, y_train)

print("Train Accuracy:", ref.score(X_train_reshaped, y_train))
print("Validation Accuracy:", ref.score(X_val_reshaped, y_val))
print("Test Accuracy :", ref.score(X_test_reshaped, y_test))

depths = [tree.tree_.max_depth for tree in ref.estimators_]
print("Min depth:", min(depths))
print("Max depth:", max(depths))
print("Average depth:", sum(depths) / len(depths))
```
