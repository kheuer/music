```python
from models import *
from features import *
from data import *
```

```python
(X_train, y_train, ids_train),(X_test, y_test, ids_test),(X_val, y_val, ids_val) = get_train_test_val_features_and_targets(df=df, feature_extractor=compute_spectral_features, splits=10, train_size=0.6, test_size=0.2)
```

```python
X_train_reshaped = X_train.reshape(X_train.shape[0], X_train.shape[1]*X_train.shape[2])
X_test_reshaped = X_test.reshape(X_test.shape[0], X_test.shape[1]*X_test.shape[2])
X_val_reshaped = X_val.reshape(X_val.shape[0], X_val.shape[1]*X_val.shape[2])
X_train_reshaped_norm = norm(X_train_reshaped, feature_extractor=compute_spectral_features)
X_test_reshaped_norm = norm(X_test_reshaped, feature_extractor=compute_spectral_features)
X_val_reshaped_norm = norm(X_val_reshaped, feature_extractor=compute_spectral_features)
```

```python
model = create_simple_feedforward_model(X_train_reshaped_norm, learning_rate=0.001)
model.summary()
```

```python
history = model.fit(
    X_train_reshaped_norm,        # Features
    y_train,                 # Integer labels
    validation_data=(X_val_reshaped_norm, y_val),
    epochs=50,
    batch_size=32,
    verbose=1
)
```

```python
test_loss, test_acc = model.evaluate(X_test_reshaped_norm, y_test, verbose=0)

print(f"Test Loss: {test_loss:.4f}")
print(f"Test Accuracy: {test_acc:.4f}")
```

```python
y_pred_prob = model.predict(X_test_reshaped_norm)
y_pred = np.argmax(y_pred_prob, axis=1)

print(y_pred[:10])
```
